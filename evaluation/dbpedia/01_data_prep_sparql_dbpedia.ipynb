```python
import json
import sys
import os

sys.path.append(os.path.abspath('..'))  # go up from dbpedia → evaluation
from utils import get_answers, save_answer_cache, answer_cache

SEMANTIC_PATH = '../../dataset/raw/jaccard_triplets_dbpedia.json'
SPLITS_PATH = '../../dataset/splits/splits_dbpedia.json'
CACHE_PATH = '../../cache/dbpedia_answer_cache.pkl'
```

```python
# Load Data
with open(SEMANTIC_PATH, 'r') as f:
    raw_semantic_data = json.load(f)
with open(SPLITS_PATH, 'r') as f:
    test_indices = json.load(f).get('test_idx', [])

unique_queries = set()
for idx in test_indices:
    item = raw_semantic_data[idx]
    if item.get('anchor_query'):
        unique_queries.add(item['anchor_query'])
        for p in item.get('positive_queries', []):
            unique_queries.add(p.get('query') if isinstance(p, dict) else p)
unique_queries = list(unique_queries)
print(f'Unique queries to cache: {len(unique_queries)}')
```

```python
for q in unique_queries:
        get_answers(q, endpoint_name="dbpedia")

os.makedirs(os.path.dirname(CACHE_PATH), exist_ok=True)
save_answer_cache(answer_cache, CACHE_PATH)
```
